```python
import numpy as np


# global constants
Rd = 287  # J kg^-1 K^-1
g = 9.81  # m s^-2
cv = 717  # J kg^-1 K^-1
cp = 1004  # J kg^-1 K^-1
kappa = Rd / cv
gamma = cp / cv


def ScaleHeight(T0):
    return (Rd * T0) / g


def NSquaredIsothermal(H):
    return (kappa * g) / H


def SoundSpeedSquared(H):
    return gamma * g * H


T0 = 275  # K
H = ScaleHeight(T0)
N_squared = NSquaredIsothermal(H)
Cs_squared = SoundSpeedSquared(H)

print(f'Scale height H={H:.3f} with T0={T0}')
print(f'N^2={N_squared:.6f} with T0={T0}')
print(f'N^2 / Cs^2 = {(N_squared / Cs_squared):.3e}')

KH_squared = N_squared / Cs_squared
Lh = (np.sqrt(2) * np.pi) / np.sqrt(KH_squared)
Lh / 1000
```
