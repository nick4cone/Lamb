```python
import numpy as np
import matplotlib.pyplot as plt
amp = 50  # hPa
halfwidth = 1 / 30
perturbation_lat = np.deg2rad(45)
perturbation_lon = np.deg2rad(180)
lat = np.deg2rad(np.arange(0, 91, step=1))
lon = np.deg2rad(np.arange(0, 361, step=1))
lon_grid, lat_grid = np.meshgrid(lon, lat)
shape = [len(lon), len(lat)]
r = np.zeros(shape)
r = np.arccos(np.sin(perturbation_lat) * np.sin(lat_grid) + np.cos(perturbation_lat) * np.cos(lat_grid) * np.cos(lon_grid - perturbation_lon))
sfc_height = np.zeros(shape)
sfc_height = amp * np.exp(-(r / halfwidth) ** 2)

fig, ax = plt.subplots()
ax.contourf(np.rad2deg(lon_grid), np.rad2deg(lat_grid), sfc_height)
```
